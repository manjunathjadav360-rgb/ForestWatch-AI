## 3. Check Sentinel-2 data for a pilot area

This is a small rectangular test area in the central Western Ghats region. The bounds are a data-access test area, not a final study boundary.

```python
aoi = ee.Geometry.Rectangle([74.5, 12.0, 75.5, 14.0])
start_date = '2025-01-01'
end_date = '2025-12-31'

sentinel2 = (ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
             .filterBounds(aoi)
             .filterDate(start_date, end_date))

count = sentinel2.size().getInfo()
print(f'Sentinel-2 scenes found: {count}')
print(f'Date window: {start_date} to {end_date}')
print('AOI bounds: [74.5, 12.0, 75.5, 14.0]')
```
